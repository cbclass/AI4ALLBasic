## 3. k-Means Clustering

Ein k-Means Modell in `scikit-learn` wird analog zum bekannten Vorgehen erstellt.

1. Erstellen eines Objekts vom Typ `KMeans` unter Angabe der relvanten Parameter:
    1. `n_clusters`:  $k$, die Anzahl der Cluster
    2. `init`: wie die initialen Clusterzentren festgelegt werden sollen. Zur Auswahl stehen `random`, `k-means++` sowie die Angabe eines Arrays mit den Koordinaten der initialen Cluster
    3. `n_init`: Anzahl der Durchläufe mit unterschiedlichen initialen Clusterzentren. Das beste Ergebnis wird zurückgegeben - gemessen an der Summe der Distanzen der Samples zum nächsten Clusterzentrum. Bei `random` defaultmäßig 10.
    4. `max_iter`: maximale Anzahl von Iterationen,  Default ist 300.
    5. `random_state`: um Clustering wiederholen zu können.
2. Trainieren des Models mit der Methods `fit()`.
3. Anwenden des trainierten Modells zur Zuordnung zu Clustern mit der Methode `predict()`.
