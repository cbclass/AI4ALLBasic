# Einführung Jupyter Notebook

Ein Notebook ist eine Art interaktive Website auf der Sie Bilder und Text zusammen mit ausführbarem Code finden. In diesem Kurs werden wir Notebooks für interaktive Übungen verwenden. 

Notebooks werden auch verwendet, um Python Code einzuführen und zu erläutern. Sie haben dann in den Notebooks die Möglichkeit, eigene Codesegmente auszuführen und Lösungen zu entwickeln.

Ein Notebook besteht aus einzelnen Elementen, sogenannten "Zellen". Diese können entweder Text oder Code beinhalten.
Ein Notebook wird erstellt, indem diese einzelne Zellen erzeugt werden. Diese können dann separat voneinander ausgeführt werden.

Bibliotheken enthalten vorgeschriebenen Code, den wir einfach nutzen können. So kann mit wenigen Zeilen Code ein gewünschtes Ergebnis erreicht werden. Heute nutzen wir die Bibliothek `pandas`. Diese vereinfacht die Arbeit in Data Science durch eine erleichterte Verwaltung von Datensätzen. Unter anderem können mit Funktionen in `pandas` auch Dateien geladen werden.
