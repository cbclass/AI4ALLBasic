## 5. Modellentwicklung mit Pipelines

### 5.1 Kreuzvalidierung - Cross Validation

Das Ergebnis des Testens eines Modells ist abhängig davon, welche Daten im Trainings- und im Testdatensatz vorhanden sind. Aus diesem Grund bietet es sich an, hier verschiedene Varianten zu testen.

Ein Verfahren hierfür ist die sogenannte **Kreuzvalidierung** oder **Cross-Validation**. Hierbei wird ein Datensatz z.B. in fünf gleich große Datensätze D1,D2,D3,D4 und D5 aufgeteilt. Diese werden als **Folds** bezeichnet. Dann werden wie folgt fünf Modelle trainiert und getestet:


| Model | Trainingsdaten | Testdaten |
|-------|----------------|-----------|
|  1    | D1, D2, D3, D4 |    D5     |
|  2    | D1, D2, D3, D5 |    D4     |
|  3    | D1, D2, D4, D5 |    D3     |
|  4    | D1, D3, D4, D5 |    D2     |
|  5    | D2, D3, D4, D5 |    D1     |

Die Funktion [cross_val_score](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html#sklearn.model_selection.cross_val_score) führt die Cross Validation durch.

Allerdings werden die Trainings- und Testdaten für jeden Schritt erneut bestimmt. Man muss, um die Unabhängigkeit der Daten sicherzustellen, die Skalierung der Daten also in jedem Schritt erneut vornehmen. Hierzu werden **Pipelines** verwendet.

In einer Pipeline definiert man eine Reihenfolge verschiedener Schritte der Datenvorverarbeitung. Alle sog. Transformer müssen die Methoden `fit()` und `transform()` implementieren. Als letztes Element kann eine Modell eingefügt werden. Mehr hierzu im [User Guide](https://scikit-learn.org/stable/modules/compose.html#pipeline).

### 5.2 Optimierung von Hyperparametern: GridSearchCV

Ein Modell kann verschiedene Hyperparameter haben, die unabhängig voneinander verschiedene Werte annehmen können. Um geeignete Hyperparameterwerte zu finden, muss man alle Kombinationen testen. 

Hier hilft **GridSearchCV**: die Kombinationen werden alle automatisch getestet und die Ergebnisse zusammengetragen. Weitere Informationen finden Sie auf [GridSearchCV](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html).

Zuerst definieren ein sogenanntes **Dictionary** mit den Parameterwerten, die wir testen wollen.
Das Dictionary with in {} eingeschlossen. Jeder potentielle Parameter wird genanngt gefogt von einer Liste der möglichen Werte (nach einem Doppelpunkt).


````python
parameter={'hidden_layer_sizes':[(10,),(20,)],'random_state':[10,20],'max_iter':[3000]}
````

Bei Erstellung der Pipeline wird ein Objekt vom `GridSearchCV` erstellt. Dieses erhält den Classifier (ohne Parameter), die Parameter, die wir definiert haben, sowie das Gütemass. Ohne weitere Angaben wird eine 5-Fold Cross Validation durchgeführt.

````python
testPipeline=Pipeline([('skalierung', StandardScaler()), 
            ('ANN', GridSearchCV(MLPClassifier(),parameter,scoring='accuracy'))])
````

Dieses Pipeline-Objekt trainieren wir nun. 

````python
testPipeline.fit(Xcancer,yCancer)
````

Und extrahieren die Ergebnisse und speichern sie in einem DataFrame ab.

````python
ergebnisse=pd.DataFrame(testPipeline["ANN"].cv_results_)
````
