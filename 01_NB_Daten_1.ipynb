## 4.  Zugriff auf Spalten des Dataframes

`pandas`  erlaubt den Zugriff Spalten eines **Dataframes** über deren Namen. Dies geschieht, indem man den Spaltennamen in eckigen Klammern angibt, z.B. `df['Spaltenname']`. Diese Methode ermöglicht es, gezielt Daten zu filtern, zu analysieren und zu manipulieren.

````python
df['Pinguinart'] # gesamte Spalte der Pinguinarten wird zurückgegeben
````

So kann auch auf mehrere Spalten gleichzeitig zugegriffen werden. 

**Hinweis:** Die **doppelten Klammern** in der folgendn Anweisung sind hier **notwendig**, weil die inneren Klammern `['Pinguinart', 'Gewicht']` eine Python-Liste definieren, die angibt welche Spalten wir haben wollen, während die äußeren Klammern `df[ ]` den eigentlichen DataFrame-Zugriff darstellen.

````python
df[['Pinguinart', 'Gewicht']] # mehrere Spalten werden zurückgegeben
````

So können die Daten auch gefiltert und z.B. Berechnungen durchgeführt werden.

### 4.1 Filterung von Daten

````python
# Filtern von Daten 
adelie_pinguine = df[df['Pinguinart'] == 'Adelie'] # Filtern der Daten nach der Pinguinart 'Adelie'
print(adelie_pinguine.shape) # 139 Pinguine gehoeren zur Art 'Adelie' und sind in der Variable 'adelie_pinguine' gespeichert, shape gibt die Anzahl der Zeilen und Spalten des DataFrames an
````

**Erläuterung:**

Mit 

```python
df['Pinguinart']=='Adelie'
```
wird jeder Wert der Spalte `Pinguinart` mit `'Adelie'` verglichen. Das Ergebnis dieses Vergleichs sind 325 Boole'sche Werte, die angeben, ob es sich bei dem Pinguin (= Sample) um einen Adeliepinguin handelt.

Mit 
```python
df[df['Pinguinart']=='Adelie']
```
greifen wir jetzt nur auf die Pinguine (Samples) zu, für die der Vergleich der Art mit Adelie den Boole'schen Wert wahr ergeben hat. Hier erhalten wir einen Dataframe mit weniger Samples, also weniger Zeilen.

Mit 
```python
adelie_pinguine=df[df['Pinguinart']=='Adelie']
```
speichern wir diesen Dataframe in einer neuen Variable ab.

Wenn wir die Dimensionen dieser Variablen ausgeben:
```python
print(adelie_pinguine.shape)
```

### 7.1 Histogramme 

Im folgenden sind die Merkmale `Insel` und `Schnabellaenge` mithilfe der Funktion `histplot()` von `seaborn` in einem Histogramm visualisiert. In beiden Fälle wurde die Anzahlder darzustellenden Balken auf 3 gesetzt (`bin=3`). 


````python
plt.close('all')

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

sns.histplot(df['Insel'], bins=3, ax=axes[0])
axes[0].set_ylabel('absolute Anzahl')

sns.histplot(data=df['Schnabellaenge'], bins=3, ax=axes[1])
axes[1].set_ylabel('absolute Anzahl')

plt.show()
````

### 7.2 Scatterplots

Mithilfe von `sns.scatterplot()` lassen sich zwei Merkmale in einem Scatterplot darstellen. Mit `sns.histplot()` lassen sich  gestapelte Histogramme darstellen. 

#### Aufgabe 12: Vergleich: Scatterplots und gestapelte Histogramme: kategorische Merkmale

Der folgende Code stellt die Beziehung zwischen Pinguinart und Heimatinsel einmal als Scatterplot und als gestapeltes Histogramm dar. Vergleichen Sie beiden Visualisierungsmethoden. Welche ist besser geeignet? Warum?

**Antwort:**

````python
plt.close('all')
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Erster Subplot: Gestapeltes Histogramm
sns.histplot(data=df, stat='count', x='Pinguinart', hue='Insel', 
             multiple='stack', ax=axes[0])
axes[0].set_title('als Histogramm')
axes[0].set_xlabel('Pinguinart')
axes[0].set_ylabel('absolute Anzahl')

# Zweiter Subplot: Scatterplot
sns.scatterplot(x=df['Pinguinart'], y=df['Insel'], ax=axes[1])
axes[1].set_title('als Scatterplot')
axes[1].set_xlabel('Pinguinart')
axes[1].set_ylabel('Insel')
plt.suptitle("Pinguinart in Bezug auf die Heimatsinsel")
plt.tight_layout()
plt.show()
````

## 8. Korrelationen
Um die in der Scatterplot-Matrix erahnbaren Zusammenhänge zwischen 
Merkmalen weiter zu untersuchen können Korrelationen untersucht werden. Sie können quantitativ auf einen Zusammenhang zwischen Merkmalen hinweisen.

### 8.1 Berechnung der Korrelationskoeffizienten

Die `pandas` Methode `df.corr()` berechnet die Korrelationskoeffizienten und stellt diese in einer Tabelle dar. Die Korrelationen können nicht für nominale Merkmale berechnet werden. Daher muss `numeric_only=True` übergeben werden, um einen Fehler zu vermeiden.

Der folgende Code berechnet die Korrelationen des Datensatzes.


````python
korrelationen = df.corr(method='pearson', numeric_only=True)
korrelationen
````

#### Aufgabe 16: Interpretation der Korrelation

### 8.2 Korrelationsmatrix

Mithilfe der Funktion `heatmap()` von seaborn lassen sich die Korrelationen in einer Korrelationsmatrix darstellen.

````python
plt.figure(figsize=(10, 8))
sns.heatmap(korrelationen, cmap="Blues", annot=True, fmt=".2f")
plt.title("Korrelation zwischen den Merkmalen", size=16)
plt.show()
````
