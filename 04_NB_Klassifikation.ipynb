## 5. Verschiedene Werte für k trainieren

Nun untersuchen wir, wie die Wahl von $k$ die Leistung des kNN-Modells beeinflusst. Die Wahl von $k$ ist entscheidend für die Balance zwischen Überanpassung (Overfitting) und Unteranpassung (Underfitting).

- Kleines $k$: Das Modell reagiert sehr stark auf einzelne Nachbarn, was zu Überanpassung führen kann. Es kann gut auf den Trainingsdaten abschneiden, aber schlecht auf den Testdaten generalisieren.
- Großes $k$: Das Modell berücksichtigt viele Nachbarn, was die Vorhersagen glättet. Ein zu großes $k$ kann dazu führen, dass wichtige lokale Strukturen ignoriert werden, was zu Unteranpassung führt.

## 7. Trainingsdaten skalieren

Der `StandardScaler` ist ein Modell, das die **z-Tranformation** implementiert. Er wird trainiert mit `fit()`. MIt dem trainierten Scaler werden Taten mit der Methode `transform()` skaliert.

#### Aufgabe 17: Skalieren von Daten
Bitte führen Sie die folgenden Codes aus, welche eine unterschideliche Vorgehensweise der Standardisierung der Trainingsdaten vornehmen.

**Variante 1: Standardisierung vor der Aufteilung**

````python
X_var1=X.copy()

# Standardisierung vor der Aufteilung
scaler1 = StandardScaler()
scaler1.fit(X_var1)
X_var1_scaled=scaler1.transform(X_var1)

# Aufteilung der Daten
X_var1_train, X_var1_test, y_var1_train, y_var1_test = train_test_split(X_var1_scaled, y, 
                                                                        test_size=0.2, 
                                                                        random_state=42)
````

````python
X_var2=X.copy()

# Aufteilung der Daten
X_var2_train, X_var2_test, y_var2_train, y_var2_test = train_test_split(X_var2, y, 
                                                                          test_size=0.2, 
                                                                          random_state=42)

# StandardScaler erstellen
scaler2 = StandardScaler()

# Scaler auf die Trainingsdaten anpassen und transformieren
scaler2.fit(X_var2_train)
X_var2_train_scaled=scaler2.transform(X_var2_train)

# Testdaten mit dem gespeicherten Scaler transformieren
X_var2_test_scaled = scaler2.transform(X_var2_test)
````
