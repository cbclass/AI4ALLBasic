## 1. Erstellen von Modellen mit `scikit-learn` - Allgemeines Vorgehen

Das Paket `sklearn` bietet Implementationen verschiedener Algorithmen des Maschinellen Lernens an. Mit diesen werden dann Modelle erstellt. 

Das Vorgehen, um Modelle zu erstellen, ist prinzipiell immer gleich.

### 1.1 Datenvorverarbeitung

Hierbei werden Fehler, Leerstellen und Duplikate behandelt. Zusätzlich müssen alle kategorischen Merkmale in numerische Merkmale umgewandelt werden, da die Methoden in `sklearn` mit kategorischen Werten nicht umgehen können.

### 1.2 Datasplit

Insbesondere bei Verfahren des supervised learnings, werden die Test- von den Trainingsdaten separiert, um eine Evaluation des Modells zu ermöglichen. Sofern Datens skaliert werden, sollte dies zuerst separat auf den Trainingsdaten erfolgen, um eine Unabhängig der Datensets zu gewährleisten. Die Parameter (z.B. Mittelwert und Standardabweichung bei der z-Transformation) werden dann gespeichert. Die Skalierung der Testdaten erfolgt dann mit diesen Parametern.

### 1.3 Erstellung und Anwendung des Modells

Die Implementationen der hier behandelten Verfahren bieten die gleichen Methoden an. Wir können daher immer nach dem folgenden Schritten vorgehen. 

1. Man erstellt ein neues Objekt eines spezifischen Modells. Hierbei werden relevante Parameter übergeben.
2. Dieses Modell wird dann trainiert, hierbei werden die Trainingsdaten, übergeben. Die Methode hierfür heißt `fit()`
3. Das Modell wird dann auf neuen Daten, den Testdaten, angewandt. Die Methode lautet `predict()`.

Daneben gibt es auch spezifische weitere Methoden, die teilweise auch mehrere Schritte kombinieren.

#### Schritt 1: Separieren des Regressanden ($y$-Werte)

Wir müssen den **Regressanden**, der abhängigen Variablen, vom **Regressor**, der unabhängigen Variablen, trennen.

Dies erfolgt mit Hilfe der Methode `pop()`. Diese Methode entfernt die Daten aus dem Dataframe, so dass wir die Daten zuerst in einen neuen Dataframe kopieren. Hierzu verwenden wir die Methode `.copy()`.  


```python
# Kopieren in einen neuen Dataframe
daten=data.copy()
# Separieren der y-Werte und speichern derselben in einer neuen Variablen
yWerte=daten.pop('y')
```

```python
# der Datenframe mit dem Regressor
daten
```

```python
# der Regressand, die y-Werte
yWerte
```

#### Schritt 2: Erstellen eines Modells für die Lineare Regression

Wir erzeugen nun ein neuen Modell des Typs `LinearRegression`. 

```python
# Erzeugung eines Objekts des Modells und Speichern in einer Variablen
regModel=LinearRegression()
```

#### Schritt 3: Trainieren des Modells

Jetzt rufen wir die Methode `fit()` auf und übergeben die $x$- und $y$-Werte, um die Regressionsgerade zu lernen. 


```python
# Lernen der Regressionsgeraden
regModel.fit(daten,yWerte)
```

#### Schritt 4: Betrachtung des Modells

Wir können nun das Modell ansehen. Hierzu gehört 
- die Ausgabe der Koeffizienten (Steigung und Intercept)
- die Bestimmung der vorhergesagten Werte $\hat{y}$
- das Plotten der Regressionsgerade

```python
# Speichern der Koeffizienten aus dem Model
steigung=regModel.coef_[0]
yAchsenabschnitt=regModel.intercept_

print(f"Die Funktion ist f(x)={steigung:.4f} x + {yAchsenabschnitt:.4f}")
```

```python
# Bestimmung der vorhergesagten Werte mit Hilfe von predict()
yPred=regModel.predict(daten)
```

```python
# Darstellung der Datenpunkte und der Regressionsgeraden
plt.plot(daten['x'],yWerte,'.')

plt.plot(daten['x'],yPred)
```

#### Schritt 5: Analyse des Modells 

Zuerst betrachten wir die Fehler (Residuals) und deren Verteilung. Hierzu können wir einen Boxplot nutzen.

```python
# Berechnung der Fehler
fehler=yWerte-yPred
plt.boxplot(fehler,vert=False,tick_labels=["Fehler"])
```

Wir betrachten die Summe des Quadratischen Fehlers sowie $R2$.

```python
# Berechnung des quadratischen Fehlers
quadFehler=((yWerte-yPred)**2).sum()
print("Der quadratische Fehler ist",quadFehler)
```

```python
# Berechnung von R2
r2=r2_score(yWerte,yPred)
print("R2 ist",r2)
```
